```python
import heapq

def solution(priorities):
    heap = []

    for priority in priorities:
        heapq.heappush(heap, priority)

    result = []

    while heap: # heap이 비어있지 않는 동안 계속 반복
        priority = heapq.heappop(heap)
        result.append(priority)

    return result

# import heapq

# def solution(priorities):
#     heap = []                          # 힙으로 쓸 리스트
#     for priority in priorities:
#         heapq.heappush(heap, priority) # heapq(모듈).heappush(heap(리스트), 값)
    
#     result = []
#     while heap:
#         priority = heapq.heappop(heap)
#         result.append(priority)
#     return result


```
