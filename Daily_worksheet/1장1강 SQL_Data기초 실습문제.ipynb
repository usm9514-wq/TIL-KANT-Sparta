## 실습 목표
- 데이터 사이언티스트가 실무에서 SQL을 사용하는 이유를 설명할 수 있다.
- Data, Database, DBMS, Query, SQL의 관계를 구조화하여 설명할 수 있다.
- 관계형 데이터베이스의 테이블 구조(Table, Row, Column, Cell)를 구분할 수 있다.
- 주요 DBMS의 특징과 PostgreSQL을 사용하는 이유를 설명할 수 있다.
- 데이터 저장 모델의 차이를 구분할 수 있다.
- SQL 작성 순서와 실행 순서의 차이를 설명할 수 있다.

## 진행 방식
- 이번 실습은 SQL 코드를 직접 실행하는 차시가 아니라 **개념 정리·비교·구조화 중심**입니다.
- 필수 실습부터 진행하고, 시간이 남으면 심화 실습을 수행하세요.
- 정답을 외우기보다 강의 내용을 바탕으로 자신의 언어로 설명하는 것이 중요합니다.

## 실습 데이터
- 별도 데이터셋 없음
- 강의에서 사용한 회원 테이블 예시를 그대로 활용합니다.

실무 시나리오

여러분은 전자상거래 서비스를 운영하는 회사의 데이터팀에 새롭게 합류했습니다. 업무를 시작하자 주문, 고객, 상품과 관련된 데이터가 데이터베이스에 저장되어 있다는 설명을 들었지만, 처음에는 데이터와 데이터베이스, DBMS, SQL이 각각 무엇을 의미하는지 구분하기 어렵습니다.

실제 데이터 분석 업무에서는 필요한 데이터를 직접 조회하기 위해 SQL을 사용하게 됩니다. 이번 실습에서는 본격적으로 SQL을 작성하기 전에 데이터가 데이터베이스에 어떤 형태로 저장되는지, SQL이 그 과정에서 어떤 역할을 하는지 기본 구조부터 이해해봅니다.
