# [1장 2강] 데이터베이스 구조 — 최종 실습문제

## 실습 목표
- 저장하려는 데이터의 성격에 맞는 데이터 타입을 선택하고 이유를 설명할 수 있다.
- 스키마의 의미를 이해하고 기본적인 테이블 구조를 설계할 수 있다.
- PK와 FK를 식별하고 각각의 규칙을 설명할 수 있다.
- 참조 무결성 위반 사례를 판단할 수 있다.
- 1NF, 2NF, 3NF의 핵심 원리를 구분할 수 있다.
- 정규화를 적용해 중복이 있는 테이블을 분리하고 장단점을 설명할 수 있다.

## 진행 방식
- 이번 실습은 실제 SQL 코드를 실행하는 차시가 아니라 **DB 구조 설계·판단·해석 중심**입니다.
- 별도 데이터셋은 사용하지 않습니다.
- 강의에 나온 작은 예시 테이블을 기준으로 문제를 해결하세요.
- 정답이 하나로 고정되지 않은 문제는 반드시 판단 근거를 함께 작성하세요.

실무 시나리오

전자상거래 서비스가 성장하면서 고객, 주문, 상품 데이터가 계속 증가하고 있습니다. 그런데 모든 정보를 하나의 테이블에 저장하거나 적절한 기준 없이 데이터를 관리하면 동일한 정보가 반복해서 저장되거나, 데이터가 서로 맞지 않는 문제가 발생할 수 있습니다.

데이터팀에서는 이러한 문제를 방지하기 위해 각 데이터에 적합한 데이터 타입을 결정하고, PK와 FK를 이용해 데이터를 식별하고 연결하며, 중복을 줄일 수 있도록 테이블 구조를 설계​해야 합니다. 이번 실습에서는 실제 데이터베이스를 설계한다는 관점에서 데이터 타입, 키, 무결성, 정규화의 기본 원리를 확인해봅니다.
