```python
counts = {}

names = ["A", "B", "A"]

for name in names:#리스트 내 모든 성분에 접근
    if name in counts:# 리스트의 현재 성분이 딕셔너리 키 값 목록에 있다면
        counts[name] += 1 #딕셔너리[키] 딕셔너리의 벨류값을 하나 증가

    else:
        counts[name] = 1

print(counts)
```

```python
def solution(participants):
    counts = {}

    for participant in participants:
        if participant in counts:      # 이미 등장한 이름이면
            counts[participant] += 1   # 횟수 +1
        else:                          # 처음 등장한 이름이면
            counts[participant] = 1    # 1로 시작

    return counts


print(solution(["minsu", "jisu", "minsu", "yuna", "minsu"]))
print(solution(["a", "b", "a", "b"]))
print(solution(["tom"]))

```
