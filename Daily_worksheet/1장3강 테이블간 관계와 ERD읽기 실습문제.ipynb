# [1장 3강] 테이블 간 관계와 ERD 읽기 — 최종 실습문제

## 실습 목표
- 1:1 / 1:N / N:M 관계 유형을 실무 예시로 구분할 수 있다.
- N:M 관계를 중간 테이블(Junction Table)을 사용해 두 개의 1:N 관계로 분리할 수 있다.
- ERD의 기본 구성 요소(Entity, Attribute, Relationship)를 설명할 수 있다.
- 까마귀발 표기법(Crow's Foot Notation)의 기호를 읽을 수 있다.
- ERD를 양방향으로 해석할 수 있다.
- 간단한 도메인의 테이블 관계를 직접 설계할 수 있다.

## 진행 방식
- 이번 실습은 SQL 실행보다 **관계 해석·ERD 읽기·설계 중심**입니다.
- 별도 데이터셋은 사용하지 않습니다.
- 강의에서 다룬 작은 예시 테이블을 기준으로 문제를 해결하세요.
- 관계 유형을 적을 때는 반드시 "양쪽이 각각 몇 개와 연결되는가"를 근거로 작성하세요.

실무 시나리오

회사 데이터베이스에는 고객, 주문, 상품처럼 서로 관련된 여러 테이블이 존재합니다. 예를 들어 고객 한 명이 여러 번 주문할 수 있고, 하나의 주문에는 여러 상품이 포함될 수 있습니다. 데이터를 정확하게 조회하려면 각각의 테이블만 보는 것이 아니라 테이블이 어떤 기준으로 연결되어 있는지 먼저 이해해야 합니다.

실무에서는 새로운 데이터베이스를 전달받거나 기존 시스템을 분석할 때 ERD를 통해 전체 데이터 구조를 파악하는 경우가 많습니다. 이번 실습에서는 1:1, 1, N:M 관계를 구분하고, PK와 FK를 바탕으로 테이블 간 관계와 ERD를 읽고 직접 설계하는 과정​을 연습합니다.
